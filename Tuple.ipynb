```python
'''
--------------------------------------------------------
Tuple :-The collection of homogeneous & hetrogeneous datatype value is called as Tuple
1...Immutable :- Never able to change in run time
2...Ordered :- Preserve insertion order but Duplicate element index shows as previous index
3...Duplicate are allowed
4...Support Indexing
5...Support Slicing

Methods of Tuple
1...Index --- Find Index of element
TupleName.index(Element) 
2...Count --- Find Frequency of element
TupleName.count(Element)
---------------------------------------------------------
Tuple Methods
│
├── count() #Find Frequency of Element
└── index() # Find Index of Element


we can able to perfrom Indexing & slicing on tuple

'''
Tuple=(1,2,3,4,5,1,2,"Laxmi")
print(type(Tuple))
print(Tuple)
print("Index :-",Tuple.index(5))#index
print("Frequency :-",Tuple.count(1))#Count
#indexing & Traversing
for Element in Tuple:
    print(Element,"----->",Tuple.index(Element))
print("Length:-",len(Tuple))
print("Address :-",id(Tuple))
```
