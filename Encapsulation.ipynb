```python
'''
Encapsultion :- The process of binding datamember & methods in single unit that single
unit is called class & that process is called Encapsulation

Class :- Class is blue print for object 
it consist of datamembers & Member Function

Object :- Object is real world entity which attribute or properties 

There are mainly 3 types of Datamember or Attributes

1..Instance :- Object Level datamember or attribute which change according to object 
Always create in constructor
inside the constructor can access datamember directly but outside constructor we have to use self parameter or object in any method
Syntax
def __init__(self,ArgumentName):     ---Constructor
    self.InstanceDatamemberName=Value ---Instance attribute declaration & Initialization
    self.InstanceDatamemberName=ArgumentName
eg ..Name , Mob , email etc 

2..Class :- Class Level datamember or attribute which are same for every object 
Always create in class & Outside the constructor
Syntax
class ClassName:
    ClassDatamember=value    ---Class attribute declaration & initialization

eg ..Course , Institute etc
3..Local :- Function & Method Level datamember 
Always create in function it can not be use outside the function
Syntax :
class ClassName:
    def Function():
        DatamemberName=value ---Declaration & Initialization of local attribute

********Methods in pythons************
There are mainly 3 types of methods in python
1..Instance 
It used for perfrom operation on instance datamember
Syntax
def MethodName(self):      ---Instance Method have first argument as self
    Body of Method

2..Class
It used for perfrom operation on instance datamember
Syntax 
@ClassMethod     ---Decorator for class method which used for identify class method
def MethodName(cls):       ---Class Method have first argument as cls
    Body of Method

3..Static
'''
class Students:
    Class="Bcs" # Class Datamembers
    def __init__(self,Name,Mob,Email): #Constructor 
        self.Name=Name
        self.Mob=Mob    #Instance Datamembers & Attributes
        self.Email=Email
    def Greet(self,Role):
        Role=Role   # Local datamember
        print(f"Your Role is {Role}")
        print("Name :-",self.Name)
        print("Class :-",self.Class)
        print("Mobile No :-",self.Mob)
        print("Email Address :-",self.Email)
S=Students("Sahebrao",9284488740,"Rahires818@gmail.com") #Object creation & passing parameter to constructor
S.Greet("Student")
    
```
