```python
'''
Abstraction :- The process of hinding low level details & complex details from general users is called as Absraction
We can achieve abstraction using Abstract methods & interface
while abstract class inheritance is must be present

Abstract Method :- The methods which always declare in parent and override or define in child that method special methods
are called as Abstract Method
it declare using following syntax 
@abstractmethod
def MethodName():
    Lines of code

Abstract Class :- The class which must contain one or more than one abstract method that class is callled
as Abstract Class
Syntax
from abc import ABC ,abstractmethod
class ClassName:
    @abstractmethod
    def MethodName():
        Lines of code

'''
from abc import ABC ,abstractmethod
class Parent:
    @abstractmethod
    def Test(self):
        pass
class Child(Parent):
    def Test(self):
        print("Hello...")
C=Child()
C.Test()
```
