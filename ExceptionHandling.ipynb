```python
'''
Exception Handling :- The process of handling error & exception while running program is called exception handling
Syntax 
try:
    Risky code
except:
    Message Related to handle risky code
finally:
    Code which want excute every time
'''
Num1=eval(input("Enter First Number...."))
Num2=eval(input("Enter Second Number..."))
try:
    print("Division :-",Num1/Num2)
except:
    print("We can`t Divide any number by zero...")
finally:
    print("Do it again....")
```
