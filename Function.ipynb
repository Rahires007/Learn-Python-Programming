```python
'''
Function :- The Reusable block of code which perfrom particular task is called Function
There are mainly 2 types of function 
1..Predefined 
eg ..print() 

2..UserDefined
Syntax
def FunctionName():
    Lines of code

There are mainly 5 ways to send the argument to function such as follow
1..Positional
2..Keyword
3..Positional arbitary argument
4..Keyword arbitary argument
5..Positional & Keyword argument 
'''
#Simple Function Without Parameter
def Greet():
    print("Hello World...")#Body of Function
Greet()

#Function With Parameter & Return
def Square(Num):
    return Num*Num #Return From Function
def Cube(Num):
    return Num*Num*Num #Return From Function

#Positional argument Passing
Square=Square(10)
print("Square :-",Square)

#Keyword Argument Passing
Cube=Cube(Num=10)
print("Cube :-",Cube)

#arbitary Positional argument passing
def Addition(*Numbers): #Collect Argument in the for tuple
    print(Numbers)
    Sum=0
    for Num in Numbers:
        Sum=Sum+Num
    return Sum
Add=Addition(10,20,30,40,50)#Sending Multiple argument in from tuple
print("Addition of Given Numbers :- ",Add)

#arbitary Keyword argument passing 
def Percentage(**Subjects):
    print(Subjects) #Collect Marks in Dictionary format
    Obtain=0
    Total=len(Subjects)*100
    for Marks in Subjects.values():
        Obtain=Obtain+Marks
    Percentage=(Obtain/Total)*100
    return Percentage
print(Percentage(S1=100,S2=90,S3=80,S4=80))

#We can send Postional & Keyword argument to single function using following Instruction
'''
1..Firstly send the positional argument 
2..After that send keyword argument
'''
#Passing positional & Keyword at same time to same function
def Add(Num1,Num2):
    print("Sum :-",Num1+Num2)
Add(10,Num2=30)
```
