```python
'''
Function Convertsto Example	      Result
int()	 Integer	   int("10")	10
float()	 Float	   float("10.5")	10.5
str()	 String	   str(100)     	"100"
list()	 List	   list((1,2,3))	[1,2,3]
tuple()	 Tuple	   tuple([1,2,3])	(1,2,3)
set()	 Set	       set([1,2,2])	{1,2}
dict()	 Dictionary dict([("a",1)])	{"a":1}
bool()	 Boolean	   bool(1)	    True
'''
Num=10
print(type(Num))
print(type(float(Num)))
print(type(str(Num)))
print(type(bool(Num)))
L=[10]
print(type(L))
print(type(tuple(L)))
print(type(set(L)))
print(type(list(tuple(L))))
```
