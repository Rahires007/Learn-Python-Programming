```python
'''
Dictionary :- The collection of homogeneous or hetrogenous element in key value pair in contigeous memory allocation is called Dictionary
1...Mutable in nature 
2...Value may be duplicate but Key Must be unique
3...Ordered 

Method of Dictionary
Dictionary Methods
│
├── keys() # Give All keys
├── values() #Give All values
├── items()  #Give All keys & Values in tuple format
├── get() # Give Values of key
├── update() #Add new key value pair
├── pop() # Remove value using specific key
├── popitem() # Pop all Recently added item
├── setdefault() 
├── clear() # clear all key value pairs
└── copy() #Copy dictionary

#Dictionary 
DictionaryName={'key':'value'}

''' 
Dictionary={'Name':'Laxmi','Age':21}
print(type(Dictionary))
print(Dictionary)
print(Dictionary.keys()) #Get all keys
print(Dictionary.values()) #Get all values
print(Dictionary.items()) # Get Element with key & value pair in the form tuple
print(Dictionary.get("Name")) #Get value of key
Dictionary.update({'City':'Nanded'}) # Update Element with key & value pair
print(Dictionary)
New=Dictionary.copy() #Copy 
print(New)
Result=Dictionary.pop('City')#Remove Value using the Key & Return that remove value
print(Result)
print(Dictionary)
Dictionary.popitem()
print(Dictionary)
Dictionary.clear()#Clear
print(Dictionary)
```
