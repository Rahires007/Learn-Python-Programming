```python
'''
All Primitive datatypes are immutable & Store single value
'''
#List :- The orderded collection homogeneous & hetrogeneous datatypes values is called as List
'''
1..Mutable --> Able to change
2..Allow Duplicate Value --> Able to insert duplicate values
3..Orderd --> Presere order of insertion , Index is preserve
'''
List=[10,20,30,40,50]
print(type(List))
print(List)
print("Address of List :-",id(List))
print("Length of List :-",len(List))#Calculate Length
List.append("Laxmi")#Add data in list
print("Count :-",List.count(10))#Count specific value 
#Traversing the list
for Val in List:
    print(Val)
#Indexing :- Access any element using index that process is called Indexing
#using indexing we can fetch one element 
print(List[0])
#Slicing :- The process access multiple element using sequence of index & step value is callled Slicing 
#Using slicing we can fetch multiple element
print("Reverse List :-",List[::-1])

print("------------------------------------------")
#Filter String values from List using isinstance function
Name_List=[]
Num_List=[]
for Name in List:
    if isinstance(Name,str):
        Name_List.append(Name)
    else:
        Num_List.append(Name)
print("Name List :-",Name_List)
print("Numeric List :-",Num_List)
print("--------------------------------------------")
```
