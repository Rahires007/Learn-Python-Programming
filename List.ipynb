```python
'''
int ,float ,string,boolean,Complex are immutable & Store single value
List,Set,Dictionary are mutable in nature
Tuple is immutable in nature
'''
#List :- The orderded collection homogeneous & hetrogeneous datatypes values is called as List
'''
---------------------------------------------------------
1..Mutable --> Able to change
2..Allow Duplicate Value --> Able to insert duplicate values
3..Orderd --> Presere order of insertion , Index is preserve
4..Support Indexing
5..Support Slicing
6..Slower than Tuple
7..Require more memory than tuple
8..Used for growing data
9..Only unpacking is possible

We can perfrom all operation on tuple using type casting
convert into List using 
ListName=List(TupleName)

#Main Important Function
1..enumerate --- It Seperate index & element of any collection or sequence like string , collection like List
Tuple 
2..Type Casting Functions

Function Convertsto Example	      Result
int()	 Integer	   int("10")	10
float()	 Float	   float("10.5")	10.5
str()	 String	   str(100)     	"100"
list()	 List	   list((1,2,3))	[1,2,3]
tuple()	 Tuple	   tuple([1,2,3])	(1,2,3)
set()	 Set	       set([1,2,2])	{1,2}
dict()	 Dictionary dict([("a",1)])	{"a":1}
bool()	 Boolean	   bool(1)	    True

3..zip --- Combine the element of collection in tuple form

------------------------------------------------------
List Methods
│
├── append() # Add Element
├── extend() # Add Multiple Element
├── insert() # Add Element at specific position or index
├── remove() # remove Element
├── pop() # Remove Last Element
├── clear() # Clear All Element
├── copy() # Copy List
│
├── index() # Find Index of Element
├── count() # Find Frequency of Element
├── sort() # Sort all Elements
└── reverse() # Reverse the List

#Create List of single element
ListName=[element]

'''
List=[10,20,30,40,50,80,90]
print(type(List))
print(List)
print("Address of List :-",id(List))
List.extend([60,70])#Extend for add multiple element at same time
print(List)
List.insert(1,10)#Insert element at specific index
print(List)
List.remove(70)#Remove element from List
print(List)
List.pop()#Remove element from List
print(List)
print("Length of List :-",len(List))#Calculate Length
List.append("Laxmi")#Add data in list
print("Count :-",List.count(10))#Count specific value 
#Traversing the list
for Val in List:
    print(Val)
#Indexing :- Access any element using index that process is called Indexing
#using indexing we can fetch one element 
print(List[0])
#Slicing :- The process access multiple element using sequence of index & step value is callled Slicing 
#Using slicing we can fetch multiple element
print("Reverse List :-",List[::-1])

print("------------------------------------------")
#Filter String values from List using isinstance function
Name_List=[]
Num_List=[]
for Name in List:
    if isinstance(Name,str):
        Name_List.append(Name)
    else:
        Num_List.append(Name)
print("Name List :-",Name_List)
print("Numeric List :-",Num_List)
List.clear()#Clear all element
print(List)
print("--------------------------------------------")
```
